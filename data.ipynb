```python
import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

n = 2500  # number of samples
stations = np.arange(1, 21)  # 20 stations

# Generate columns
station_id = np.random.choice(stations, n)
time_of_day = np.random.randint(0, 24, n)
day_of_week = np.random.randint(0, 7, n)
is_holiday = np.random.binomial(1, 0.08, n)
weather = np.random.choice(['Sunny', 'Rainy', 'Cloudy'], n, p=[0.6, 0.2, 0.2])
nearby_events = np.random.poisson(0.4, n)
temperature = np.random.normal(30, 5, n).round(1)
platform_capacity = np.random.choice([300, 500, 700, 900], n)

# Compute a synthetic "crowd score"
score = (
    (np.sin(time_of_day / 24 * 2 * np.pi) + 1) * 2
    + (day_of_week / 6)
    + nearby_events * 0.8
    + (is_holiday * 2.5)
    + (platform_capacity / 900)
    + (weather == 'Rainy') * 1.2
)

# Convert score into crowd_level categories
quantiles = np.quantile(score, [0.33, 0.66])
crowd_level = np.where(score <= quantiles[0], 'Low',
               np.where(score <= quantiles[1], 'Medium', 'High'))

# Assemble the DataFrame
df = pd.DataFrame({
    'station_id': station_id,
    'time_of_day': time_of_day,
    'day_of_week': day_of_week,
    'is_holiday': is_holiday,
    'weather': weather,
    'nearby_events': nearby_events,
    'temperature': temperature,
    'platform_capacity': platform_capacity,
    'crowd_level': crowd_level
})

# Save to CSV
df.to_csv("metro_crowd_data.csv", index=False)

print("✅ Dataset saved as metro_crowd_data.csv")
print(df.head())
```
